## Skill Ratings Heatmap


```python
#| echo: false

import pandas as pd
import plotly.graph_objects as go

# Build the DataFrame
data = {
    "Skill": ["AWS Cloud", "Python", "SQL", "R", "Tableau", "Excel", "PySpark", "PowerBI", "GitHub", "Hadoop"],
    "Bowen":   [2, 4, 4, 3, 3, 5, 1, 5, 3, 1],
    "Mahnoor": [4, 4, 3, 2, 5, 5, 4, 5, 4, 3],
    "Feidhelm": [3, 4, 3, 3, 3, 4, 2, 5, 4, 1]
}

df = pd.DataFrame(data)

# Prepare inputs for the heatmap
skills = df["Skill"].tolist()
people = ["Bowen", "Mahnoor", "Feidhelm"]
z_values = df[people].values

# Create Plotly heatmap
fig = go.Figure(
    data=go.Heatmap(
        z=z_values,
        x=people,
        y=skills,
        colorbar=dict(title="Rating")
    )
)

fig.update_layout(
    title="Skill Ratings Heatmap",
    xaxis_title="Person",
    yaxis_title="Skill"
)

fig  # display in Quarto
```
